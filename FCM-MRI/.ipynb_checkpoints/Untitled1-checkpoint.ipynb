```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import cv2
import importlib
import bcifswsni_bn
importlib.reload(bcifswsni_bn)

from bcifswsni_bn import run_model
```

```python
img = cv2.imread('test.png', 0)
img = cv2.resize(img, (80,80)) # grayscale

seg, U, V, B = run_model(img, 5, 1.3, 1, 'yager', 0.1, p=2)
```

```python
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title("Original")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(seg, cmap='viridis')
plt.title("Segmented")
plt.axis('off')
```

```python
seg_img = np.zeros_like(img, dtype=float)

for i in range(U.shape[0]):   # clusters
    seg_img[seg == i] = i * (255 // U.shape[0])

plt.imshow(seg_img, cmap='gray')
plt.title("Segmented Image (Reconstructed)")
plt.axis('off')
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(img, cmap='gray')
plt.title("Original")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(seg, cmap='viridis')
plt.title("Cluster Labels")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(seg_img, cmap='gray')
plt.title("Segmented Image")
plt.axis('off')
```
