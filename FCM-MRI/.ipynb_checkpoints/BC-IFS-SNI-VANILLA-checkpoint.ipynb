```python
def Calculate_lambda(A,U,V,m):
    n = np.size(A,axis=0)
    c = np.size(V,axis=0)
    d = np.size(V,axis=1)

    um = U**m

    term1 = (um@V[:,:,0])/(um.sum(axis=1)[:,None] + 1e-8)
    term1 = A[:,:,0] - term1

    term1 = term1.sum(axis=0)   #check if axis=0
    term2 = um.sum(axis=1)
    term3 = (1/(term2 + 1e-8)).sum()

    return 2*term1/term3
```

```python
def Update_B(A,V,U,m):
    term1 = (um@V[:,:,0])/(um.sum(axis=1)[:,None] + 1e-8)
    term1 = A[:,:,0] - term1
    term2 = Calculate_lambda(A,U,V,m)/(2*(um.sum(axis=1)[:,None] + 1e-8))
    return term1-term2
    
```

```python
def Update_Y(A,B,g,beta=None,p=None):
    n = np.size(A,axis=0)
    d = np.size(A,axis=1)
    Y = np.zeros((n,d,3))
    Y[:,:,0] = A[:,:,0] - B
    mu = Y[:,:,0]
    if g == "yager":
        Y[:,:,1] = (1 - mu**p)**(1/p)
    else-if g=="sugeno":
        Y[:,:,1] = 1-mu/(1+beta*mu)

    Y[:,:,2] = 1-Y[:,:,0]-Y[:,:,1]

    return Y
    
```
