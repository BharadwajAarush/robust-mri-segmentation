```python
def Calculate_Denominator_for_U(A,B,V,m,a_,N,j):
    c = np.size(V,axis=0)
    val = 0
    for i in range(c):
        temp = Calculate_Sum_for_U(A,B,V,m,a_,N,i,j)
        val = val + temp**(-1/m-1)
    
        
    
```

```python
def Update_U(A,B,V,m,a_,N):
    n = np.size(A,axis=0)
    c = np.size(V,axis=0)

    U = np.eros((n,c))
    for i in range(n):
        for j in range(c):
            num = Calculate_Sum_for_U(A,B,V,m,a_,N,i,j)
            den = Calculate_Denominator_for_U(A,B,V,m,a_,N,j):
            U[i][j] = (num**(-1/m-1))/den
    
    
```
